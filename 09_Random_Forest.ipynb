## **1. Ensemble Learning**

**Understand the Concept**

- Ensemble Learning combines multiple models to make one prediction.
- It can improve the overall performance of the model.
- Random Forest is an Ensemble Learning method.
- It uses multiple Decision Trees together.

**How It Works**

- Multiple models are trained using the data.
- Each model makes a prediction.
- The predictions are combined to get the final result.

**Example**

- Several Decision Trees predict employee Attrition.
- The combined prediction gives the final Attrition result.

## **2. Bagging**

**Understand the Concept**

- Bagging means Bootstrap Aggregating.
- It creates multiple models using different samples of the training data.
- Each model makes its own prediction.
- The predictions are combined to get a final prediction.

**How It Works**

- Different samples are created from the training data.
- A Decision Tree is trained on each sample.
- All trees make predictions.
- The predictions are combined using voting.

**Example**

- Different Decision Trees are trained using different employee samples.
- Their predictions are combined to predict Attrition.

## **3. Random Forest**

**Understand the Concept**

- Random Forest is an Ensemble Learning algorithm.
- It combines multiple Decision Trees.
- Each tree learns from different samples and features.
- The final prediction is based on the predictions of all trees.

**How It Works**

- Multiple Decision Trees are created.
- Each tree gets a random sample of the data.
- Each tree uses a random selection of features.
- The predictions from all trees are combined.

**Example**

- 100 Decision Trees predict employee Attrition.
- The majority prediction becomes the final result.

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("employee_attrition_decision_tree.csv")

X = df[["Age","Monthly_Income","Job_Satisfaction","Years_at_Company","Work_Hours",
        "Job_Level","Distance_from_Home","Training_Hours"]]
y = df["Attrition"]

model = RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(X, y)
print("Random Forest model trained successfully.")
print("Number of Trees:", model.n_estimators)
```

**Explanation**

- Random Forest uses multiple Decision Trees.
- n_estimators=100 creates 100 trees.
- All trees make predictions and combine them.
- The combined result gives the final prediction.

## **4. Multiple Decision Trees**

**Understand the Concept**

- Random Forest uses many Decision Trees instead of one tree.
- Each tree learns from different samples of the data.
- Each tree makes its own prediction.
- The predictions are combined to make the final result.

**How It Works**

- Random Forest creates multiple Decision Trees.
- Each tree gives a prediction.
- The predictions are combined using voting.
- The final prediction is based on the majority vote.

## **5. Random Feature Selection**

**Understand the Concept**

- Random Forest does not use all features for every tree.
- It randomly selects a smaller group of features.
- This makes the trees different from each other.
- It helps improve the overall model performance.

**How It Works**

- A random group of features is selected at each split.
- The tree finds the best feature from that group.
- Different trees use different feature combinations.
- This helps reduce overfitting.


```python
print("Random Feature Selection applied.")
print("Features considered at each split:", model.max_features)
```

## **6. Voting**

**Understand the Concept**

- Voting combines the predictions from multiple Decision Trees.
- Each tree gives its own prediction.
- The class with the most votes becomes the final prediction.
- It is mainly used for Classification.

## **7. Feature Importance**

**Understand the Concept**

- Feature Importance shows how useful each feature is for making predictions.
- Higher importance means the feature has more influence on the model.
- Random Forest calculates importance for all input features.
- It helps us understand which features are most useful.

**How It Works**

- Each Decision Tree checks different features.
- Random Forest combines their importance values.
- Features with higher values are considered more important.

**Example**

- Job_Satisfaction may have higher importance than Training_Hours.
- This means Job_Satisfaction has more influence on the prediction.

```python
importance = pd.DataFrame({"Feature": X.columns,"Importance": model.feature_importances_})
importance = importance.sort_values("Importance",ascending=False)
print(importance)
```

**Explanation**

- feature_importances_ gives the importance of each feature.
- Higher value means more important feature.
- Sorting helps us see the most important features first.

## **8. Overfitting**

**Understand the Concept**

- Overfitting happens when the model learns the training data too closely.
- The model performs well on training data but poorly on new data.
- Random Forest reduces overfitting by combining many Decision Trees.
- Using multiple trees makes the model more stable.

**How It Works**

- Each tree learns from different samples and features.
- The trees make different predictions.
- Random Forest combines these predictions to reduce the effect of one tree.

**Example**

- One Decision Tree may overfit the training data.
- Random Forest combines many trees to reduce overfitting.

## **9. Advantages**

**Understand the Concept**

- Random Forest is more stable than a single Decision Tree.
- It can handle large datasets.
- It can be used for Classification and Regression.
- It can reduce overfitting.
- It can show Feature Importance.



## **10. Limitations**

**Understand the Concept**

- Random Forest can be slower than a single Decision Tree.
- It can use more memory because it contains many trees.
- The model can be harder to understand.
- Training many trees can take more time.



## **11. Decision Tree vs Random Forest**

**Understand the Concept**

- Decision Tree uses one tree to make predictions.
- Random Forest combines multiple Decision Trees.
- Random Forest usually reduces the effect of overfitting.
- Decision Tree is easier to understand and visualize.

| Feature | Decision Tree | Random Forest |
|---|---|---|
| Number of Trees | One | Multiple |
| Complexity | Simple | More complex |
| Overfitting | More likely | Less likely |
| Speed | Faster | Usually slower |
| Interpretation | Easy | More difficult |
| Feature Importance | Available | Available |
| Prediction | One tree | Combined predictions |

**Example**

- Decision Tree → One tree makes the final prediction.
- Random Forest → Multiple trees vote to make the final prediction.

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("employee_attrition_decision_tree.csv")

X = df[["Age","Monthly_Income","Job_Satisfaction","Years_at_Company","Work_Hours",
        "Job_Level","Distance_from_Home","Training_Hours"]]
y = df["Attrition"]

decision_tree = DecisionTreeClassifier(random_state=42)
random_forest = RandomForestClassifier(n_estimators=100,random_state=42)
decision_tree.fit(X, y)
random_forest.fit(X, y)
print("Decision Tree trained.")
print("Random Forest trained.")
print("Random Forest Trees:", random_forest.n_estimators)
```

**Explanation**

- Decision Tree uses one tree.
- Random Forest uses many trees.
- Random Forest combines the predictions of all trees.
- Both can be used for classification and regression.
