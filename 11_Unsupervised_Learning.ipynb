# **Notebook 11 – Unsupervised Learning**

## **1. What is Unsupervised Learning**

**Understand the Concept**

- Unsupervised Learning works with data that has no labelled target.
- The model finds patterns or groups in the data by itself.
- There is no correct output given during training.
- It is useful when labelled data is unavailable.

**How It Works**

- The model receives input data without labels.
- It looks for patterns, similarities, or relationships.
- The model creates groups or reduces the number of features.


## **2. Supervised vs Unsupervised Learning**

**Understand the Concept**

- Supervised Learning uses labelled data with a known target.
- Unsupervised Learning uses data without a target label.
- Supervised Learning learns to predict the given target.
- Unsupervised Learning finds patterns or groups by itself.

**How It Works**

- Supervised → Input data + Target → Prediction.
- Unsupervised → Input data → Patterns or Groups.


## **3. Clustering**

**Understand the Concept**

- Clustering is an Unsupervised Learning method.
- It groups similar data points together.
- Data points in the same group are more similar to each other.
- There is no target label given to the model.

**How It Works**

- The model checks similarities between data points.
- It creates different groups called clusters.
- Each cluster contains similar data points.


```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
df = pd.read_csv("customer_segmentation.csv")
X = df[["Age","Annual_Income","Spending_Score"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
model = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X_scaled)
print(df[["Age", "Annual_Income", "Spending_Score", "Cluster"]].head())
```

**Explanation**

- KMeans creates 3 customer groups.
- StandardScaler puts the features into a similar range.
- Cluster shows the group assigned to each customer.

## **4. Dimensionality Reduction**

**Understand the Concept**

- Dimensionality Reduction reduces the number of features in a dataset.
- It keeps the most useful information from the data.
- It makes the data simpler and easier to visualize.
- PCA is a common Dimensionality Reduction method.

**How It Works**

- The model looks at the relationships between features.
- It creates new features called Principal Components.
- These components represent the important information from the original data.



## **5. Applications**


- Unsupervised Learning is used to find hidden patterns in data.
- It can be used for customer segmentation.
- It can help detect unusual or different data points.
- It is useful for exploring data when labelled data is unavailable.
